```python
import numpy as np
x = np.array([1,2,3,4])
y = np.array([2,3,4,5])
sum_x = sum(x)
sum_y = sum(y)
sum_xy = sum(x*y)
sum_x2 = sum(x**2)
sum_y2 = sum(y**2)
n = len(x)
r = (n * sum_xy - sum_x * sum_y) / (np.sqrt((n * sum_x2 - sum_x**2) * (n * sum_y2 - sum_y**2)))
print(f"The Correlation Coefficient is: {r:.4f}")

x_mean = np.mean(x)
y_mean = np.mean(y)
covariance = np.sum((x - x_mean) * (y - y_mean)) / (n - 1)
print(f"The Covariance is: {covariance:.4f}")
```
