```python
import numpy as np
import pandas as pd

def get_frequent_itemsets_df(data, min_support=0.6):
    transactions = [set(t) for t in data.values()]
    N = len(transactions)
    min_count = min_support * N
    unique_items = sorted(list(set(item for t in transactions for item in t)))
    current_candidates = [frozenset([item]) for item in unique_items]
    records = []
    while current_candidates:
        frequent_k = []
        for candidate in current_candidates:
            count = sum(1 for t in transactions if candidate.issubset(t))
            if count >= min_count:
                support = count / N
                frequent_k.append(candidate)
                records.append({
                    "Itemset Size": len(candidate),
                    "Support": support,
                    "Frequent Itemset": f"{{{', '.join(sorted(list(candidate)))}}}"
                })
        if not frequent_k:
            break
        next_candidates = set()
        for i in range(len(frequent_k)):
            for j in range(i + 1, len(frequent_k)):
                itemset1 = sorted(list(frequent_k[i]))
                itemset2 = sorted(list(frequent_k[j]))
                if itemset1[:-1] == itemset2[:-1]:
                    next_candidates.add(frequent_k[i] | frequent_k[j])
        current_candidates = list(next_candidates)
    df = pd.DataFrame(records)
    df["Support"] = (df["Support"] * 100).astype(int).astype(str) + "%"
    df = df.sort_values(by=["Itemset Size", "Frequent Itemset"]).reset_index(drop=True)
    return df
dataset = {
    "T1" : ["Bread", "Milk"],
    "T2" : ["Bread", "Diaper", "Beer", "Eggs"],
    "T3": ["Milk", "Diaper", "Beer", "Coke"],
    "T4": ["Bread", "Milk", "Diaper", "Beer"],
    "T5": ["Bread", "Milk", "Diaper", "Coke"]
}

frequent_df = get_frequent_itemsets_df(dataset, min_support=0.6)
print(frequent_df.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from scipy import stats

columns = ["Strawberry", "Coffee", "Orange", "Vanilla"]
data = {"Male":[23,18,8,8], "Female":[15,6,12,10]}

df = pd.DataFrame(data, index=columns)
chi2, p_value, dof, expected = stats.chi2_contingency(df, correction=False)

print("--- a. Hypotheses ---")
print("Null Hypothesis (H0): Chocolate cream flavor preference is INDEPENDENT of gender.")
print("Alternative Hypothesis (H1): Chocolate cream flavor preference is DEPENDENT on gender.\n")
print("--- b. Degrees of Freedom ---")
print(f"Degrees of freedom (df): {dof}\n")
print("--- c. Expected Frequency Table ---")
df_expected = pd.DataFrame(expected, index=columns, columns=data.keys())
print(df_expected.round(2),"\n")
print("--- Full Statistical Summary ---")
print(f"Chi-Square Statistic (X²): {chi2:.4f}")
print(f"p-value: {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print(f"\nConclusion: Reject H0 at alpha = {alpha}. Flavor preference is dependently associated with gender.")
else:
    print(f"\nConclusion: Fail to reject H0 at alpha = {alpha}. There is no significant evidence that flavor preference depends on gender.")
```
