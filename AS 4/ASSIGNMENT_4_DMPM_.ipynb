```python
import pandas as pd
data= pd.read_csv("Student_Marks.csv")
print(data)
```

```python
import matplotlib.pyplot as plt
plt.plot(data["Marks"])
plt.title("Line Chart")
plt.show()
```

```python
plt.bar(data.index,data["number_courses"])
plt.title("Bar Plot")
plt.show()
```

```python
import seaborn as sns
plt.boxplot(data["number_courses"])
plt.title("Box Plot")
plt.show()
```

```python
course_count = data["number_courses"].value_counts()
plt.pie(course_count, labels=course_count.index, autopct="%1.1f%%")

plt.title("Number of Courses Distribution")

plt.show()
```

```python
plt.scatter(data["time_study"], data["Marks"], data["number_courses"],color="Red")
plt.title("Scatter Plot")
plt.show()
```

```python
plt.hist(data["Marks"])
plt.title("Histogram")
plt.show()
```
