```python
import numpy as np
import pandas as pd

def get_frequent_itemsets_df(data, min_support=0.6):
    transactions = [set(t) for t in data.values()]
    N = len(transactions)
    min_count = min_support * N
    unique_items = sorted(list(set(item for t in transactions for item in t)))
    current_candidates = [frozenset([item]) for item in unique_items]
    records = []
    while current_candidates:
        frequent_k = []
        for candidate in current_candidates:
            count = sum(1 for t in transactions if candidate.issubset(t))
            if count >= min_count:
                support = count / N
                frequent_k.append(candidate)
                records.append({
                    "Itemset Size": len(candidate),
                    "Support": support,
                    "Frequent Itemset": f"{{{', '.join(sorted(list(candidate)))}}}"
                })
        if not frequent_k:
            break
        next_candidates = set()
        for i in range(len(frequent_k)):
            for j in range(i + 1, len(frequent_k)):
                itemset1 = sorted(list(frequent_k[i]))
                itemset2 = sorted(list(frequent_k[j]))
                if itemset1[:-1] == itemset2[:-1]:
                    next_candidates.add(frequent_k[i] | frequent_k[j])
        current_candidates = list(next_candidates)
    df = pd.DataFrame(records)
    df["Support"] = (df["Support"] * 100).astype(int).astype(str) + "%"
    df = df.sort_values(by=["Itemset Size", "Frequent Itemset"]).reset_index(drop=True)
    return df
dataset = {
    "T1" : ["Bread", "Milk"],
    "T2" : ["Bread", "Diaper", "Beer", "Eggs"],
    "T3": ["Milk", "Diaper", "Beer", "Coke"],
    "T4": ["Bread", "Milk", "Diaper", "Beer"],
    "T5": ["Bread", "Milk", "Diaper", "Coke"]
}

frequent_df = get_frequent_itemsets_df(dataset, min_support=0.6)
print(frequent_df.to_string(index=False))
```
