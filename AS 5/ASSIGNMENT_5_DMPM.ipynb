```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

# Z-Score
mean = np.mean(data)
std = np.std(data)

z = (data - mean) / std
print("Z-Score:", z)

# Min-Max
minimum = np.min(data)
maximum = np.max(data)

mm = (data - minimum) / (maximum - minimum)
print("Min-Max:", mm)

# Decimal Scaling
ds = data / 100

print("Decimal Scaling:", ds)
```

```python
import pandas as pd

data = pd.read_csv("Student_Marks.csv")

print("Original Data")
print(data.head())

# Decimal Scaling

data["number_courses"] = data["number_courses"] / 10
data["time_study"] = data["time_study"] / 10
data["Marks"] = data["Marks"] / 100

print("\nAfter Decimal Scaling")
print(data.head())
```

```python
import pandas as pd

data = pd.read_csv("Student_Marks.csv")

print("Original Data")
print(data.head())

# Min-Max Normalization

data["number_courses"] = (data["number_courses"] - data["number_courses"].min()) / (data["number_courses"].max() - data["number_courses"].min())

data["time_study"] = (data["time_study"] - data["time_study"].min()) / (data["time_study"].max() - data["time_study"].min())

data["Marks"] = (data["Marks"] - data["Marks"].min()) / (data["Marks"].max() - data["Marks"].min())

print("\nAfter Min-Max Normalization")
print(data.head())
```

```python
import pandas as pd

data = pd.read_csv("Student_Marks.csv")

print("Original Data")
print(data.head())

# Z-Score Normalization

data["number_courses"] = (data["number_courses"] - data["number_courses"].mean()) / data["number_courses"].std()

data["time_study"] = (data["time_study"] - data["time_study"].mean()) / data["time_study"].std()

data["Marks"] = (data["Marks"] - data["Marks"].mean()) / data["Marks"].std()

print("\nAfter Z-Score Normalization")
print(data.head())

print("\nMean")
print(data.mean())

print("\nVariance")
print(data.var())

print("\nStandard Deviation")
print(data.std())
```
